```python
import pandas as pd
import numpy as np
import xgboost as xgb
import matplotlib.pyplot as plt
import os
import sys
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
# --- Constants from Competition Rules ---
# --- Constants from Competition Rules ---
T_COST_BIPS = 2
T_COST_PERCENT = T_COST_BIPS * 0.0001  # 0.0002 or 0.02%
MIN_TRADE_GAP = 15  # in seconds (which is 1 row)
EOD_PENALTY_PCT = 0.01  # 1%
STARTING_CAPITAL = 100000  # <--- ADD THIS LINE

# --- Feature Selection (from your Quant HP1.pdf) ---
# A much more detailed list based on your PDF snippets.
# --- Feature Selection (Curated list from Quant HP1.pdf) ---
# --- Feature Selection (Using your specific 10-feature list) ---
SELECTED_FEATURES = [
    'PB9_T1', 'PB11_T1', 'PB10_T1', 'PB11_T2', 'PB10_T2', 
        'PB11_T3', 'PB10_T3', 'PB11_T4', 'PB10_T4', 'PB11_T5', 'PB10_T5',
        'PB11_T6', 'PB10_T6', 'PB11_T7', 'PB10_T7', 'PB11_T8', 'PB10_T8',
        'PB11_T9', 'PB10_T9', 'PB11_T10', 'PB10_T10', 'PB11_T11', 'PB10_T11',
        'PB11_T12', 'PB10_T12', 'PB13_T12', 
    'VB3_T3', 'VB4_T3', 'VB5_T7', 'VB2_T3', 'VB5_T9', 'VB5_T8', 'VB2_T11', 'VB3_T4', 'VB4_T4', 'VB2_T4', 'VB3_T5', 'VB4_T5', 'VB3_T6', 'VB4_T6', 'VB2_T10', 'VB2_T5', 'VB3_T11', 'VB4_T11', 'VB2_T6', 'VB3_T7', 'VB4_T7', 'VB3_T10', 'VB4_T10', 'VB3_T8', 'VB4_T8', 'VB2_T9', 'VB2_T7', 'VB2_T8', 'VB3_T9', 'VB4_T9'
,'BB5_T1', 'BB13_T1', 'BB4_T1', 'BB6_T1', 'BB14_T1', 'BB15_T1', 'BB11_T1',
    'BB12_T1', 'BB13_T2', 'BB5_T2', 'BB7_T1', 'BB8_T1', 'BB14_T2', 'BB15_T2', 'BB4_T2', 'BB6_T2',
    'BB11_T2', 'BB12_T2', 'BB7_T2', 'BB8_T2', 'BB5_T3', 'BB13_T3', 'BB14_T3', 'BB15_T3', 'BB4_T3', 
    'BB6_T3', 'BB12_T3', 'BB11_T3', 'BB8_T3', 'BB7_T3', 'BB13_T4', 'BB5_T4', 'BB14_T4', 'BB15_T4', 'BB6_T4', 'BB4_T4', 'BB12_T4', 'BB5_T5', 'BB13_T5', 'BB14_T5'


]
# Columns to load from the CSVs
# We only load what we absolutely need.
COLUMNS_TO_LOAD = ['Time', 'Price'] + SELECTED_FEATURES

```

```python

def load_data(file_path):
    """
    Loads a single day's CSV, loading only the necessary columns.
    """
    if not os.path.exists(file_path):
        print(f"Error: File not found at {file_path}")
        print("Please make sure your data is in a 'data/' folder and named day_1.csv, day_2.csv, etc.")
        return None
    
    try:
        # usecols is critical for performance
        df = pd.read_csv(file_path, usecols=COLUMNS_TO_LOAD)
        return df
    except ValueError as e:
        print(f"Error loading {file_path}. A feature might be missing.")
        print(f"Details: {e}")
        print("Please check if all SELECTED_FEATURES are in your CSVs.")
        return None


def create_labels(df, look_ahead=3, threshold_pct=0.00000005):
    """
    Creates the target variable (y) based on the "look ahead" strategy.
    
    Args:
        df (pd.DataFrame): The dataframe with price data.
        look_ahead (int): How many seconds (rows) to look ahead for the signal.
        threshold_pct (float): The minimum % change needed to generate a signal.
                               0.0005 = 0.05% or 5 Bips.
    
    Returns:
        pd.Series: A series with signals {-1, 0, 1}.
    """
    
    # Calculate future price
    future_price = df['Price'].shift(-look_ahead)
    
    # Calculate percentage change
    pct_change = (future_price - df['Price']) / df['Price']
    
    # Create signals
    signals = pd.Series(0, index=df.index)
    signals.loc[pct_change > threshold_pct] = 1
    signals.loc[pct_change < -threshold_pct] = -1
    
    # Fill NaN (at the end of the dataframe) with 0
    signals = signals.fillna(0).astype(int)
    # Assuming 'signals' is a Pandas Series
    print("the 1's are", signals.value_counts().get(1, 0))
    # Assuming 'signals' is a Pandas Series
    print("the 0's are", signals.value_counts().get(0, 0))
    # Assuming 'signals' is a Pandas Series
    print("the -1's are", signals.value_counts().get(-1, 0))
    return signals
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

def train_model(X_train, y_train, params):
    """
    Trains a robust logistic regression model using a pipeline
    that includes feature scaling.
    """
    print("Training Logistic Regression pipeline...")
    
    # 1. Create the pipeline
    #    Step 1: StandardScaler() - Scales all features
    #    Step 2: LogisticRegression() - The linear model
    pipe = Pipeline([
        ('scaler', StandardScaler()),
        ('model', LogisticRegression(multi_class='auto', max_iter=1000, random_state=42))
    ])
    
    # 2. Set the parameters from Optuna
    #    Note: The parameters are for the *model* step in the pipeline.
    pipe.set_params(**params)
    
    # 3. Fit the entire pipeline on the training data
    pipe.fit(X_train, y_train)
    
    print("Model training complete.")
    
    # The returned 'model' is actually the entire fitted pipeline
    return pipe
```

```python
import optuna

def objective(trial, X_train, y_train, test_files, selected_features):
    """
    The main objective function for Optuna.
    It trains a Logistic Regression pipeline and runs a full backtest, 
    returning the annualized return.
    """
    
    # 1. Define the hyperparameter search space for LogisticRegression
    #    We use 'model__<param_name>' to set params for the pipeline step
    params = {
        'model__C': trial.suggest_float('model__C', 1e-4, 1e2, log=True),
        'model__penalty': trial.suggest_categorical('model__penalty', ['l1', 'l2']),
        'model__solver': 'saga' # 'saga' solver supports both l1 and l2
    }
    
    # 2. Train the model (pipeline)
    try:
        model = train_model(X_train.copy(), y_train.copy(), params)
    except Exception as e:
        print(f"Trial failed during training: {e}")
        return -1e9  # Return a very bad score if training fails

    # 3. Run the full backtest loop (this part is unchanged)
    all_final_pnls = []
    all_max_drawdowns = []

    for i, f_path in enumerate(test_files):
        day_num = 2 + i 
        
        # --- IMPORTANT: Ensure you are loading your data correctly ---
        # (Using the load_data function you already have)
        test_df = load_data(f_path) 
        if test_df is None:
            continue
        
        test_df['day'] = day_num
        
        # run_backtest now receives the *pipeline* as the 'model'
        # It will automatically scale the test data before predicting
        daily_pnel_series = run_backtest(test_df, model, selected_features)
        
        # Calculate metrics
        cum_ret, mdd, daily_calmar = calculate_metrics(daily_pnel_series)
        
        all_final_pnls.append(cum_ret)
        all_max_drawdowns.append(mdd)

    # 4. Calculate final objective metric
    annualized_return = 0.0
    
    if all_final_pnls:
        avg_daily_return = np.mean(all_final_pnls)
        annualized_return = avg_daily_return * 252 # 252 trading days

        # --- OPTUNA PRUNING (Optional but recommended) ---
        trial.report(annualized_return, 0)
        if trial.should_prune():
            raise optuna.exceptions.TrialPruned()

    # 5. Return the metric to be MAXIMIZED
    # Handle case where no pnl was generated
    if not all_final_pnls:
        return 0.0

    return annualized_return
```

```python
def run_backtest(daily_df, model, feature_cols):
    """
    Runs a full, row-by-row backtest that respects all competition rules
    and calculates PnL based on a fixed STARTING_CAPITAL.
    """
    
    # This series will store the PERCENTAGE PnL, e.g., (total_pnl / 100000)
    pnl_series = pd.Series(0.0, index=daily_df.index)
    
    position = 0  # Can be -1, 0, 1
    last_trade_time = -MIN_TRADE_GAP
    entry_price = 0.0
    
    # --- NEW: We track PnL in DOLLARS, not percent ---
    total_absolute_pnl = 0.0
    last_price = daily_df['Price'].iloc[0]
    
    was_penalized = False # Flag for EOD penalty

    # Get all predictions for the day at once
    all_features = daily_df[feature_cols]
    predicted_signals_mapped = model.predict(all_features)
    predicted_signals = pd.Series(predicted_signals_mapped).replace({0: -1, 1: 0, 2: 1}).values

    for i in range(len(daily_df)):
        
        row = daily_df.iloc[i]
        price = row['Price']
        
        # 1. Calculate PnL change (in dollars) from the *previous* second
        pnl_change_dollars = 0.0
        if position == 1:
            pnl_change_dollars = (price - last_price)  # PnL for 1 unit
        elif position == -1:
            pnl_change_dollars = (last_price - price)  # PnL for 1 unit (inverted)
        
        total_absolute_pnl += pnl_change_dollars
        
        predicted_signal = predicted_signals[i]
        
        # 2. Check for End-of-Day (EOD) square-off
        is_last_second = (i == len(daily_df) - 1)
        if is_last_second and position != 0:
            predicted_signal = 0  # Force a "flat" signal to close
            was_penalized = True

        # 3. Check for trade cooldown
        can_trade = (i - last_trade_time) >= MIN_TRADE_GAP
        
        # 4. Execute Trade Logic
        if can_trade:
            # --- GOING LONG ---
            if predicted_signal == 1 and position == 0:
                position = 1
                entry_price = price
                last_trade_time = i
                # Apply T-cost in DOLLARS
                total_absolute_pnl -= (price * T_COST_PERCENT) 

            # --- GOING SHORT ---
            elif predicted_signal == -1 and position == 0:
                position = -1
                entry_price = price
                last_trade_time = i
                # Apply T-cost in DOLLARS
                total_absolute_pnl -= (price * T_COST_PERCENT)
                
            # --- CLOSING LONG ---
            elif (predicted_signal == 0 or predicted_signal == -1) and position == 1:
                position = 0
                last_trade_time = i
                # Apply T-cost in DOLLARS
                total_absolute_pnl -= (price * T_COST_PERCENT)
                
            # --- CLOSING SHORT ---
            elif (predicted_signal == 0 or predicted_signal == 1) and position == -1:
                position = 0
                last_trade_time = i
                # Apply T-cost in DOLLARS
                total_absolute_pnl -= (price * T_COST_PERCENT)

        # 5. Record the PERCENTAGE PnL for this second
        # This is the final value used for all metrics
        pnl_series.iloc[i] = total_absolute_pnl / STARTING_CAPITAL
        
        last_price = price

    # 6. Apply EOD Penalty (as a percentage, e.g., 0.01)
    if was_penalized:
        pnl_series.iloc[-1] -= EOD_PENALTY_PCT
        
    return pnl_series
```

```python
def calculate_metrics(pnl_series):
    """
    Calculates key performance metrics from a cumulative PnL series.
    """
    # 1. Cumulative Return (is just the last value)
    cumulative_return = pnl_series.iloc[-1]
    
    # 2. Maximum Drawdown (MDD)
    # We track the highest "peak" PnL we've seen so far.
    # The drawdown is the difference between that peak and the current PnL.
    running_peak = 0.0
    max_drawdown = 0.0
    
    for pnl_value in pnl_series:
        # Update the peak if we have a new high
        if pnl_value > running_peak:
            running_peak = pnl_value
        
        # Calculate the drawdown from the current peak
        drawdown = running_peak - pnl_value # This is a positive number
        
        # Update the maximum drawdown seen so far
        if drawdown > max_drawdown:
            max_drawdown = drawdown
            
    # Avoid division by zero if PnL is always negative
    if cumulative_return == 0 and max_drawdown == 0:
        calmar_ratio = 0.0
    elif max_drawdown == 0:
        calmar_ratio = np.inf # Or some large number, as there's no drawdown
    else:
        # Note: This is a DAILY Calmar. We will annualize in main.
        calmar_ratio = cumulative_return / max_drawdown

    return cumulative_return, max_drawdown, calmar_ratio
```

```python
def main():
    # --- 1. Load and Prepare Training Data (Do this ONCE) ---
    print("Loading training data...")
    train_files = [f'data\\day{i}.csv' for i in range(18)]
    
    # Your test_files list from the original main function
    test_files = [
        'data\\day18.csv', 'data\\day19.csv', 'data\\day20.csv', 
        'data\\day21.csv', 'data\\day22.csv'
    ]

    all_train_df = []
    for f_path in train_files:
        df = load_data(f_path)
        if df is not None:
            all_train_df.append(df)
            
    if not all_train_df:
        print("Error: No training data loaded. Exiting.")
        return

    train_df = pd.concat(all_train_df, ignore_index=True)
    
    print("Creating training labels...")
    X_train, y_train = create_labels(train_df, SELECTED_FEATURES)
    print(f"Total training rows: {len(X_train)}")
    
    if X_train.empty:
        print("Error: No training data generated. Exiting.")
        return

    # --- 2. Define and Run Optuna Study ---
    print("\n--- Starting Optuna Hyperparameter Tuning ---")
    
    # We want to MAXIMIZE the annualized_return
    study = optuna.create_study(direction='maximize') 
    
    # We need to pass the training data and test files to the objective
    # We use a lambda function to "wrap" the objective
    objective_with_data = lambda trial: objective(trial, X_train, y_train, test_files)
    
    # Start the optimization!
    # n_trials=50 is a good start. Increase this for better results.
    study.optimize(objective_with_data, n_trials=50)

    # --- 3. Print the Results ---
    print("\n--- Optuna Tuning Complete ---")
    
    print(f"Best trial number: {study.best_trial.number}")
    print(f"Best Annualized Return: {study.best_value * 100:.4f}%")
    
    print("\nBest Parameters:")
    for key, value in study.best_params.items():
        print(f"  {key}: {value}")

    # --- 4. (Optional) Train final model and run backtest to plot ---
    print("\n--- Running final backtest with best parameters ---")
    best_params = study.best_params
    final_model = train_model(X_train, y_train, best_params)
    
    plt.figure(figsize=(15, 10))
    plt.title('Strategy PnL (Cumulative %) - Best Model', fontsize=16)
    
    # ... (Copied from your original main) ...
    all_final_pnls = []
    all_max_drawdowns = []

    for i, f_path in enumerate(test_files):
        day_num = 2 + i 
        print(f"Running backtest on {f_path} (Day {day_num})...")
        test_df = load_data(f_path[:8] + f_path[9:])
        if test_df is None:
            continue
        
        test_df['day'] = day_num
        daily_pnl_series = run_backtest(test_df, model, SELECTED_FEATURES)
        cum_ret, mdd, daily_calmar = calculate_metrics(daily_pnl_series)
        
        all_final_pnls.append(cum_ret)
        all_max_drawdowns.append(mdd)
        
        print(f"Day {day_num} -> Cum. Return: {cum_ret * 100:.4f}% | Max Drawdown: {mdd * 100:.4f}%")
        label = (f'Day {day_num} | PnL: {cum_ret*100:.2f}% | MDD: {mdd*100:.2f}%')
        plt.plot(daily_pnl_series.index, daily_pnl_series.values, label=label)

    plt.xlabel('Time (seconds)')
    plt.ylabel('Cumulative PnL (%)')
    plt.legend()
    plt.grid(True, linestyle='--', alpha=0.5)
    plt.tight_layout()
    plt.savefig('best_model_backtest_pnl.png')
    print("\nBacktest complete. Plot saved to 'best_model_backtest_pnl.png'")

    # ... (Final metrics calculation from original main) ...
    if all_final_pnls:
        print("\n--- Summary Statistics (Best Model) ---")
        avg_daily_return = np.mean(all_final_pnls)
        print(f"Average Daily Return: {avg_daily_return * 100:.4f}%")
        
        annualized_return = avg_daily_return * 252
        print(f"Annualized Return: {annualized_return * 100:.4f}%")
        
        avg_max_drawdown = np.mean(all_max_drawdowns)
        print(f"Average Max Drawdown: {avg_max_drawdown * 100:.4f}%")
        
        if avg_max_drawdown == 0:
            calmar_ratio = np.inf
        else:
            calmar_ratio = annualized_return / avg_max_drawdown
        
        print(f"--- CALMAR RATIO (Annualized): {calmar_ratio:.4f} ---")


if __name__ == '__main__':
    main()
```
